```python
import re
def remove_punctuation(text):
  return re.sub(r"[^\w\s]","",text)
df['review']=df['review'].apply(remove_punctuation)
```

```python
df['review'] = df['review'].apply(lambda x: re.sub(r'\s+', ' ', x).strip())
```

```python
df['review'] = df['review'].apply(lambda x: re.sub(r'\d+', '', x))
```

```python
df['clean_review'] = df['review']
display(df.head())
```

```python
def remove_urls(text):
  return re.sub(r'https?://\s+|www\.\s+','',text)
df['clean_review']=df['clean_review'].apply(remove_urls)
```

```python
def remove_emails(text):
    return re.sub(r'\S*@\S*\s?', '', text)
df['clean_review'] = df['clean_review'].apply(remove_emails)
display(df.head())
```

```python
def remove_html_tags(text):
    return re.sub(r'<.*?>', '', text)
df['clean_review'] = df['clean_review'].apply(remove_html_tags)
display(df.head())
```

```python
def remove_special_characters(text):
    return re.sub(r'[^a-zA-Z0-9\s]', '', text)
df['clean_review'] = df['clean_review'].apply(remove_special_characters)
display(df.head())
```

```python
def remove_emojis(text):
    emoji_pattern = re.compile("["
                               u"\U0001F600-\U0001F64F"
                               u"\U0001F300-\U0001F5FF"
                               u"\U0001F680-\U0001F6FF"
                               u"\U0001F1E0-\U0001F1FF"
                               u"\U00002702-\U000027B0"
                               "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)
df['clean_review'] = df['clean_review'].apply(remove_emojis)
display(df.head())
```

```python
df.drop('review', axis=1, inplace=True)
display(df.head())
```

```python
df['clean_txt_advance'] = df['clean_review']
display(df.head())
```

```python
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    return ' '.join([word for word in str(text).split() if word not in stop_words])

df['clean_review'] = df['clean_review'].apply(remove_stopwords)
display(df.head())
```

```python
import re
slang_dict = {
    'r': 'are',
    'gr8': 'great',
    'u': 'you',
    'ur': 'your',
    'lol': 'laughing out loud',
    'brb': 'be right back',
    'btw': 'by the way'
}

def replace_slang(text, slang_dictionary):
    for slang, replacement in slang_dictionary.items():
        text = re.sub(r'\b' + re.escape(slang) + r'\b', replacement, text, flags=re.IGNORECASE)
    return text
df['clean_txt_advance'] = df['clean_txt_advance'].apply(lambda x: replace_slang(x, slang_dict))

display(df.head())
```
